```python
import json
import csv

# ---------- input / output paths ----------
INPUT_JSON = "/Users/sherrypan/GitHub/GAR_SKGQA/datasets/dblp/test/answers.json"
OUTPUT_CSV = "/Users/sherrypan/GitHub/GAR_SKGQA/datasets/dblp/project_data/_answers.csv"

# ---------- helper function ----------
def parse_answer(answer_obj):
    """
    Distinguish between SELECT and BOOLEAN queries
    and return the answer as a dictionary.
    """
    # BOOLEAN (ASK) query
    if "boolean" in answer_obj:
        return {"boolean": answer_obj["boolean"]}

    # SELECT query
    if "results" in answer_obj and "head" in answer_obj:
        vars_ = answer_obj["head"].get("vars", [])
        bindings = answer_obj["results"].get("bindings", [])

        # If multiple rows exist, collect values as lists
        result = {var: [] for var in vars_}

        for row in bindings:
            for var in vars_:
                if var in row:
                    result[var].append(row[var].get("value"))

        # Simplify single-value lists
        for var in result:
            if len(result[var]) == 1:
                result[var] = result[var][0]

        return result

    # Fallback (unexpected structure)
    return {}

# ---------- main ----------
def main():
    with open(INPUT_JSON, "r", encoding="utf-8") as f:
        data = json.load(f)

    rows = []

    for item in data.get("answers", []):
        qid = item.get("id")
        answer_obj = item.get("answer", {})

        parsed_answer = parse_answer(answer_obj)

        rows.append({
            "id": qid,
            "answer": json.dumps(parsed_answer, ensure_ascii=False)
        })

    # Write CSV
    with open(OUTPUT_CSV, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["id", "answer"])
        writer.writeheader()
        writer.writerows(rows)

    print(f"CSV written to: {OUTPUT_CSV}")

# ---------- run ----------
if __name__ == "__main__":
    main()
```

```python
import csv
import json
from collections import defaultdict

# ------------------ helpers ------------------

def parse_ground_truth(answer_str):
    """
    Parse ground truth JSON answer into:
    - set of URIs (SELECT)
    - boolean (ASK)
    """
    obj = json.loads(answer_str)

    if "boolean" in obj:
        return obj["boolean"]

    ans = obj.get("answer", [])
    if isinstance(ans, list):
        return set(ans)
    return {ans}


def parse_prediction(pred_str):
    """
    Parse generated output result into a set of strings.
    Split on newlines and commas, but NOT on comma-space (", ").
    This handles:
    - Newline-separated: "item1\nitem2" -> ["item1", "item2"]
    - Comma-separated: "item1,item2" -> ["item1", "item2"]
    - Entity names with commas: "King Saud University, Saudi Arabia" -> ["King Saud University, Saudi Arabia"]
    """
    import re
    
    if not pred_str or pred_str.strip() == "":
        return set()

    items = []
    # Split on newlines first
    for line in pred_str.splitlines():
        line = line.strip()
        if not line:
            continue
        
        # Split on comma that is NOT followed by a space
        # This preserves entity names like "King Saud University, Saudi Arabia"
        parts = re.split(r',(?!\s)', line)
        
        for part in parts:
            part = part.strip()
            if part:
                items.append(part)

    return set(items)


def precision_recall_f1(gt, pred):
    if not pred:
        return 0.0, 0.0, 0.0

    tp = len(gt & pred)
    precision = tp / len(pred) if pred else 0.0
    recall = tp / len(gt) if gt else 1.0
    f1 = 0.0 if precision + recall == 0 else 2 * precision * recall / (precision + recall)

    return precision, recall, f1


# ------------------ load ground truth ------------------

ground_truth = {}
with open("/Users/sherrypan/GitHub/GAR_SKGQA/datasets/dblp/project_data/test_answers.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        ground_truth[row["id"]] = parse_ground_truth(row["answer"])
        # print(f"the ground truth for {row['id']} is {ground_truth[row['id']]}")


# ------------------ load hypothesis (ignore extra columns) ------------------

predictions = {}
questions = {}

with open("/Users/sherrypan/GitHub/GAR_SKGQA/results/dblp/dblp/answer_generation_Meta-Llama-3.1-8B-Instruct.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    required_cols = {"id", "generated_sparql"}
    missing = required_cols - set(reader.fieldnames)
    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    for row in reader:
        qid = row["id"]
        # replace the "" at the beginning or end of the generated_sparql
        predictions[qid] = parse_prediction(row["generated_sparql"].strip('"'))
        questions[qid] = row.get("question", "")


# ------------------ evaluation ------------------

# store both per-question metrics and aggregate lists
per_question_metrics = {}
metrics = defaultdict(list)
boolean_correct = 0
boolean_total = 0
unanswered = 0

for qid, gt in ground_truth.items():
    pred = predictions.get(qid, set())
    
    

    if not pred:
        unanswered += 1

    # BOOLEAN (ASK)
    if isinstance(gt, bool):
        boolean_total += 1
        if isinstance(pred, bool) and pred == gt:
            boolean_correct += 1
        continue

    # SELECT
    p, r, f1 = precision_recall_f1(gt, pred)
    if qid == "Q0007":
        print(f"the ground truth for {qid} is {gt}")
        print(f"the prediction for {qid} is {pred}")
        print(f"For question {qid}, precision: {p}, recall: {r}, f1: {f1}")
    metrics["precision"].append(p)
    metrics["recall"].append(r)
    metrics["f1"].append(f1)
    per_question_metrics[qid] = {"precision": p, "recall": r, "f1": f1}
    
# add the metrics to the lists and save to a new column in the original csv



# ------------------ reporting ------------------

def avg(xs):
    return sum(xs) / len(xs) if xs else 0.0

print("=== SELECT queries ===")
print(f"Macro Precision: {avg(metrics['precision']):.4f}")
print(f"Macro Recall:    {avg(metrics['recall']):.4f}")
print(f"Macro F1:        {avg(metrics['f1']):.4f}")

print(f"Answered rate:   {1 - unanswered / len(ground_truth):.4f}")
print(f"unanswered: {unanswered}")

if boolean_total > 0:
    print("\n=== BOOLEAN queries ===")
    print(f"Accuracy: {boolean_correct / boolean_total:.4f}")
    print(f"boolean correct: {boolean_correct}, boolean total: {boolean_total}")
```

```python
import csv

GROUND_TRUTH_CSV = "/Users/sherrypan/GitHub/GAR_SKGQA/datasets/dblp/project_data/test_answers.csv"
HYPOTHESIS_CSV = "/Users/sherrypan/GitHub/GAR_SKGQA/results/dblp/dblp/answer_generation_Meta-Llama-3.1-8B-Instruct.csv"
OUTPUT_CSV = "/Users/sherrypan/GitHub/GAR_SKGQA/results/dblp/dblp/answer_comparison_Meta-Llama-3.1-8B-Instruct.csv"

# Load ground truth
ground_truth = {}
with open(GROUND_TRUTH_CSV, newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        ground_truth[row["id"]] = row  # keep entire row as-is

# Load hypothesis (only need id, question, generated_sparql)
hypothesis = {}
with open(HYPOTHESIS_CSV, newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        hypothesis[row["id"]] = {
            "question": row.get("question", ""),
            "hypothesis_answer": row.get("generated_sparql", "")
        }

# Merge into comparison CSV, also add the metrics
with open(OUTPUT_CSV, "w", newline="", encoding="utf-8") as f:
    # include metric columns so DictWriter accepts extra keys
    fieldnames = ["id", "question", "hypothesis_answer", "ground_truth_answer", "precision", "recall", "f1"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    i = 0

    for qid, gt_row in ground_truth.items():
        hyp_row = hypothesis.get(qid, {"question": "", "hypothesis_answer": ""})
        metric_row = per_question_metrics.get(qid, {"precision": "", "recall": "", "f1": ""})
        writer.writerow({
            "id": qid,
            "question": hyp_row["question"],
            "hypothesis_answer": hyp_row["hypothesis_answer"],
            "ground_truth_answer": gt_row.get("answer", ""),
            "precision": metric_row.get("precision", ""),
            "recall": metric_row.get("recall", ""), 
            "f1": metric_row.get("f1", ""),
        })
        i += 1

print(f"Comparison CSV written to: {OUTPUT_CSV}")
```
